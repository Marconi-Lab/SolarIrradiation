```python
from susse import irradiation
```

```python
kampala = irradiation.Location(latitude=0.347594, longitude=32.58210, timezone='Africa/Kampala')
```

```python
estimator = irradiation.IrradianceEstimator(kampala)
```

```python
pipeline = irradiation.IrradiancePipeline(estimator)
```

```python
result = pipeline.run(
    start_date='2023-01-01',
    end_date='2023-01-03',
    cloud_cover_fraction=0.2,
    surface_tilts=[0],
    surface_azimuth=180
)

result
```
