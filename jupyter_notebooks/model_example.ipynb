```python
from pvlib.location import Location
import pandas as pd

kampala = Location(latitude=0.347594, longitude=32.58210)
start_date='2023-01-01'
end_date='2023-01-03'
freq = "1h"
time = pd.date_range(
            start=start_date, end=end_date, freq=freq, tz=kampala.tz
        )
clear_sky_est = kampala.get_clearsky(times=time, model="simplified_solis")
```
