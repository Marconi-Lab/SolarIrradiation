```python
from susse import irradiation
kampala = irradiation.Location(latitude=0.347594, longitude=32.58210, timezone='Africa/Kampala')
```

```python
kampala.altitude
```

```python
estimator = irradiation.IrradianceEstimator(kampala)
```

```python
pipeline = irradiation.IrradiancePipeline(estimator)
```

```python
pipeline.run(
    start_date='2023-01-01',
    end_date='2023-01-03',
    cloud_cover_fraction=0.2,
    surface_tilts=[0],
    surface_azimuth=180
)
```

```python
print(f"times: {type(pipeline.times)}\n"
      f"clearsky_irradiance: {type(pipeline.clear_sky)}\n"
      f"adjusted_ghi: {type(pipeline.adjusted_ghi)}\n"
      f"solar_position: {type(pipeline.solar_position)}\n"
      f"dhi: {type(pipeline.dhi)}\n"
      f"dni: {type(pipeline.dni)}\n"
      f"poa: {type(pipeline.poa_irradiance)}\n"
)
```

```python
pipeline.solar_position.head()
```

```python
estimator.decompose_irradiance(
    ghi=pipeline.adjusted_ghi,
    solar_zenith=pipeline.solar_position["zenith"].values,
    times=pipeline.times
    )
```
